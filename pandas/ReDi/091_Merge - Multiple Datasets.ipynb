```python
import pandas as pd

# Syntax to Merge Three DataFrames
# Example DataFrames
df1 = pd.DataFrame({
    'key': ['A', 'B', 'C'],
    'value_df1': [1, 2, 3]
})

df2 = pd.DataFrame({
    'key': ['B', 'C', 'D'],
    'value_df2': [4, 5, 6]
})

df3 = pd.DataFrame({
    'key': ['C', 'D', 'E'],
    'value_df3': [7, 8, 9]
})

# Merging three DataFrames on 'key'
merged_df = pd.merge(df1, df2, on='key', how='inner')
merged_df = pd.merge(merged_df, df3, on='key', how='inner')

# Display the final merged DataFrame
print(merged_df)
```

```python
# Syntax to Merge Three DataFrames in a Single Expression
import pandas as pd

# Example DataFrames
df1 = pd.DataFrame({
    'key': ['A', 'B', 'C'],
    'value_df1': [1, 2, 3]
})

df2 = pd.DataFrame({
    'key': ['B', 'C', 'D'],
    'value_df2': [4, 5, 6]
})

df3 = pd.DataFrame({
    'key': ['C', 'D', 'E'],
    'value_df3': [7, 8, 9]
})

# Merging three DataFrames on 'key' in one line
merged_df = pd.merge(pd.merge(df1, df2, on='key', how='inner'), df3, on='key', how='inner')

# Display the final merged DataFrame
print(merged_df)
```
