## Selcting Rows
`.loc()` and `.iloc()` select rows wherby:
- .loc selects by label (what's printed on the screen), while 
- .iloc selects by integer position (0-indexed location, like standard Python lists)

A label can be an integer which makes it sometimes hard to distinguish from a positional index.

```python
# select row by index
df.iloc[0]

#select by row label
df_indexed.loc["CA-2013-152156"]
```

Confusion arises if the deliberatly set index is the same a the default position. For example the first row of your data frame has the default index position 0 and the index you set also starts at zero, and your index is an integer. There is no visible difference between the two, but they are very different because label names are not positional indices.

```python
# selcting row by label or index position yields the same only for default index
df.loc[0] == df.iloc[0]
```

Another trap is a label based index that looks like a number but actually is a string. 

```python
df_postalindex = df.set_index(str("Postal Code"))
# df_postalindex.iloc[42420] # won't work
df_postalindex.loc[42420].head(2) # will work because the index is now a string type, not an integer
```

## Selecting Scalars

The `at[]` and `iat[]` functions are designed for high-speed access to a single values.

```python
df.head(1)
```

```python
# at[] — by label
df.at[0, "Sales"] = 999        # set value at row 0, column "Sales"
df.at[0, "Sales"]              # get value at row 0, column "Sales"
```

```python

# iat[] — by position
df.iat[0, 3] = "12-11-2013"    # set value at row 0, column position 3
df.iat[0, 3]                   # get value at row 0, column position 3
```

```python
## Boolean Indexing & Advanced Filtering
```

```python
df[~(df["Sales"] < 5000.0)]
```

```python
# boolean filter: Orders with Sales greater than $500
high_sales_mask = df["Sales"] > 1500.0
df[high_sales_mask].head(3)
```

Boolean filtering with multiple conditions using logical operators: AND (&), OR (|), NOT (~).

Enclose each condition in parentheses due to Python operator precedence.

combine filters or negate conditions with ~

#to answer e.g. who are the customers in the South region with Profit greater than $1500, but not in the Corporate segment?

```python
complex_filter = (df["Region"] == "South") & (df["Profit"] > 1500) & ~ (df["Segment"]== "Corporate")

df[complex_filter]
```

```python
# check membership filtering with .isin()
region_mask = df["Region"].isin(["East", "West"])
df[region_mask].head(3)
```

```python
# string accessor boolean evaluation (.str.startswith, .str.contains)
subcat_mask = df["Product Name"].str.contains("tool") | df["City"].str.contains("Berlin")
df[subcat_mask]
```
