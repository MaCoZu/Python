```python
df["Order Date"] = pd.to_datetime(df["Order Date"], format="%d-%m-%Y", errors="coerce")
df["Ship Date"] = pd.to_datetime(df["Ship Date"], format="%d-%m-%Y", errors="coerce")
```

```python
df["Order Date"].min()
df["Order Date"].max()

df["Ship Date"].min()
df["Ship Date"].max()
```

```python
df[df["Order Date"]=="2014-05-05"]
df[(df["Order Date"]=="2014-05-05") & (df.Category=="Furniture")]
df[(df["Order Date"]=="2014-05-05") & (df.Region.isin(["South", "West"]))]
```

```python
df.groupby("Region")["Profit"].mean()
```

```python
df['Cost'] = df['Sales'] - df['Profit']
df
```
