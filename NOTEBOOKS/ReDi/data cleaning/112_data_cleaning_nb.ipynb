#### backward & forward fill

```python
# Create a range of dates
dates = pd.date_range(start="2024-01-01", periods=7, freq="D")

weather = pd.DataFrame({
    "Date": dates,
    "City": ["Berlin"] * 7,
    "Temp_Celsius": [2.1, np.nan, 3.5, np.nan, np.nan, 1.8, 2.0],
})

weather
```

```python
weather["Temp_Celsius_ffill"] = weather["Temp_Celsius"].ffill() # fill na with values that came before - fill forward
weather["Temp_Celsius_bfill"] = weather["Temp_Celsius"].bfill() # fill backward
weather["Temp_Celsius_bfill_l"] = weather["Temp_Celsius"].bfill(limit=1) # how far forward shoul the function look for a value to impute
weather
```

## Dates

```python
df_date = pd.DataFrame({
    "Customer_ID": [1, 2, 3, 4, 5],
    "Signup_Date": ["2023-01-15", "15/01/2023", "Jan 20, 2023", "2023.01.25", None],
    "Last_Login": [
        "2024-03-01 14:30:40",
        "2024-03-02 09:15:60",
        "2024-03-05 22:00:10",
        "2024-03-07 12:00:05",
        "2024-03-10 08:00:19",
    ],
    "Birth_Year": [1990, 1985, 2000, 1978, 1995],
})
```

```python
df_date.info()
```

```python
# make column a date
df_date["Last_Login"] = pd.to_datetime(df_date["Last_Login"])
df_date["Last_Login"]
```

```python
df_date.info()
```

```python
# 'mixed' parameter detects all kinds of formats and converts them to proper dates in one column
df_date["Signup_Date"] = pd.to_datetime(df_date["Signup_Date"], format="mixed")
```

```python
# df_date.info()
df_date
```

```python
# access date functions with .dt
# df_date["Signup_Date"].dt.day_of_week
df_date["Signup_Date"].dt.day_name()
```

```python
# How long has it been since they signed up?
reference_date = pd.Timestamp("2024-04-01") # make reference a date for datetime calulations
df_date["Account_Age"] = reference_date - df_date["Signup_Date"]
df_date["Account_Age_Days"] = df_date["Account_Age"].dt.days # get days only
df_date
```

```python
# Calculate difference between logins
df_date["Inactivity_Period"] = df_date["Last_Login"].diff()  # Difference from the previous row
df_date
```

```python
# Keep full timestamp (Datetime)
df_date["Last_Login_Full"] = pd.to_datetime(df_date["Last_Login"])

# Extract just the date (Date)
df_date["Last_Login_Date"] = df_date["Last_Login_Full"].dt.date
df_date
```

```python
df_date["Last_Login_Full"].dt.floor("min") # you can shorten dates to hours (h), minutes (min), seconds (s)
```

```python
# Convert 1990 (int) to 1990-01-01 (datetime) - format="%Y" tells the function what your passing in
df_date["Birth_Date"] = pd.to_datetime(df_date["Birth_Year"], format="%Y")
```

## Replace

```python
df_replace = pd.DataFrame({
    "Ticket_ID": [1, 2, 3, 4, 5, 6],
    "Status": ["Open!", "In Progress", "Closed", "None", "Pending", "In Progress"],
    "Phone Number": ["555-1234", "555.5678", "!Unknown", "555-9999", "N/A", "555 0000"],
    "Feedback": ["Great!", "Bad...", "very happy", "Terrible!!!", "Happy", "Excellent service"],
    "Costs": ["$3.28", "$5.5", "$8.41", "$1.2", "$9.76", "4.83"]
})
df_replace
```

```python
df_replace["Status"].unique() # check nunique values
```

```python
df_replace.replace('!', '', regex=True) # replace all ! from the df
```

```python
# Replace specific defined in a list strings with a one standard value
df_replace["Status"] = df_replace["Status"].replace(["None", "Pending"], "Open")
df_replace
```

```python
# change column substrings in a column header
# .str accesses the string and look for pattern inside, regex=False tells the function not to treat regex symbols (^*[]()) as code
df_replace.columns = df_replace.columns.str.replace(' Number', '', regex=False)
df_replace
```

```python
# Using a dictionary for multiple different replacements at once
df_replace["Status"] = df_replace["Status"].replace({"In Progress": "Active", "Closed": "Resolved"})

df_replace
```

```python
# here we have no .str but regex=True reaches into the strings and searches for substrings (parts of strings)
# r'\$' is a regex patters looking for dollar signs inside the strings
df_replace.replace(
    {
    'Feedback': {'!': '', 'Great': 'Good', "Terrible": "Bad", },
    'Costs': {r'\$': ''},
    },
regex=True)
```

```python
df_replace
```

```python
# standardize "very happy" and "Happy" to "Good"
# (?i) is the regex flag for "ignore case"
df_replace.replace({'Feedback': {r'(?i)happy': 'Good'}}, regex=True)
```

```python
# Replace any non-digit (except the start) with a dash
# r'[^\d]' is a regex pattern for 'everything that is not a digit'
# it turns 555.5678 and 555 0000 into 555-5678 and 555-0000
df_replace['Phone'] = df_replace['Phone'].replace(r'[^\d]', '-', regex=True)
df_replace
```

```python
# .str accessor by defalut look inside the string no regex is needed here
df_replace.columns = df_replace.columns.str.replace(' ', '_')
df_replace
```

```python
# reintroduce 'dirty' NA values just to clean them in the next cell
df_replace.loc[0, 'Status'] = 'Unknown'
df_replace.loc[1, 'Phone'] = 'N/A'
df_replace
```

```python
# Turn placeholder strings into actual NaN objects
df_replace.replace(['Unknown', 'N/A', "---", "-------"], np.nan)
```
